```python
# Convergence Plot
fig, ax = plt.subplots(figsize=(8, 4.5), dpi=150)
ax.plot(trials_df["number"], trials_df["value"], marker="o", color="#2563eb", alpha=0.5, label="Trial Score")
ax.plot(trials_df["number"], trials_df["value"].cummax(), color="#dc2626", lw=2, label="Best Cumulative F1")
ax.set_title("Optuna Bayesian Optimization Convergence (MLP)", fontsize=11, fontweight="bold")
ax.set_xlabel("Trial Number")
ax.set_ylabel("Validation Macro F1")
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend()
plt.tight_layout()
plt.show()
```
